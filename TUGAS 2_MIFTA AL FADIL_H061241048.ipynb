## **1. Prompt R — Role (Identitas dan kompetensi AI)**

```python

import matplotlib.pyplot as plt

kategori = ["Geofisika", "Geolistrik", "Wenner"]
nilai = [1, 2, 3]

plt.figure(figsize=(8, 5))
plt.plot(kategori, nilai, marker="o", linewidth=2)
plt.title("Prompt R - Role")
plt.xlabel("Bidang Kajian")
plt.ylabel("Urutan Fokus Kajian")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## **2. Prompt T — Task (Tugas yang dikerjakan)**

```python

import matplotlib.pyplot as plt

tahapan = [
    "Input Data",
    "Faktor Geometri",
    "Resistivitas Semu",
    "Interpretasi"
]
urutan = [1, 2, 3, 4]

plt.figure(figsize=(9, 5))
plt.plot(tahapan, urutan, marker="o", linewidth=2)
plt.title("Prompt T - Task")
plt.xlabel("Tahapan Perhitungan")
plt.ylabel("Urutan Tahapan")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## **3. Prompt C — Context (Konteks ilmiah)**

```python

import matplotlib.pyplot as plt

material = [
    "Tanah",
    "Batuan Lapuk",
    "Zona Jenuh Air"
]
resistivitas = [50, 150, 30]

plt.figure(figsize=(8, 5))
plt.plot(material, resistivitas, marker="o", linewidth=2)
plt.title("Prompt C - Context")
plt.xlabel("Material Ilustratif")
plt.ylabel("Resistivitas (ohm.m)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## **4. Prompt I — Input (Data pengukuran)**

```python

import matplotlib.pyplot as plt

parameter = [
    "Spasi (m)",
    "Arus (mA)",
    "Beda Potensial (mV)"
]
data = [10, 200, 15.4]

plt.figure(figsize=(8, 5))
plt.bar(parameter, data)
plt.title("Prompt I - Data Pengukuran")
plt.ylabel("Nilai Input")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## **5. Prompt K — Constraints (Batasan dan validasi)**

```python

import matplotlib.pyplot as plt

parameter = [
    "Spasi > 0",
    "Arus > 0",
    "Tegangan >= 0"
]
status = [1, 1, 1]

plt.figure(figsize=(8, 5))
plt.bar(parameter, status)
plt.title("Prompt K - Validasi Data")
plt.ylabel("Status (1 = Valid)")
plt.ylim(0, 1.3)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## **6. Prompt O — Output (Hasil dalam narasi akademik)**

```python

import math
import matplotlib.pyplot as plt

# Data contoh tugas
a = 10.0
arus_mA = 200.0
potensial_mV = 15.4

# Konversi satuan
arus_A = arus_mA / 1000
potensial_V = potensial_mV / 1000

# Perhitungan Wenner
K = 2 * math.pi * a
rho_a = K * potensial_V / arus_A

# Grafik hasil
kategori = [
    "Faktor Geometri K (m)",
    "Resistivitas Semu (ohm.m)"
]
hasil = [K, rho_a]

plt.figure(figsize=(8, 5))
batang = plt.bar(kategori, hasil)

plt.title("Prompt O - Hasil Perhitungan Geolistrik")
plt.ylabel("Nilai")
plt.grid(axis="y", linestyle="--", alpha=0.5)

for bar in batang:
    tinggi = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        tinggi,
        f"{tinggi:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

Catatan: Grafik Prompt O menghitung nilai dari data tugas, sedangkan grafik R, T, I, dan K memvisualisasikan kategori atau tahapan. Grafik C memakai nilai ilustratif. Hanya grafik O yang menampilkan hasil perhitungan resistivitas semu secara fisik dari data contoh tersebut.
