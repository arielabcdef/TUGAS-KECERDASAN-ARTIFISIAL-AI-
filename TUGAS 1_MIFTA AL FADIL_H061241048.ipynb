## **MODUL: MENYUSUN PROMPT AI YANG BAIK UNTUK PERMASALAHAN GEOFISIKA**
## **FRAMEWORK R–T–C–I–K–O**

<br>

</div>



## **1. Mengapa Prompt yang Baik Sangat Penting di Geofisika?**

AI generatif (seperti Grok, ChatGPT, Claude, Gemini) dapat membantu menghitung faktor geometri, resistivitas semu, membuat interpretasi awal, bahkan menyusun laporan. Namun hasilnya sangat bergantung pada **kualitas prompt**.

Prompt yang buruk biasanya menghasilkan:
- Perhitungan salah satuan
- Interpretasi litologi yang terlalu berani tanpa data pendukung
- Jawaban berbentuk daftar poin yang tidak akademis
- Asumsi data yang tidak pernah diberikan

Prompt yang baik memaksa AI untuk:
- Mengikuti alur ilmiah yang benar
- Menghormati batasan data
- Menghasilkan narasi akademik yang runtut
- Menjelaskan asumsi secara eksplisit

Modul ini mengajarkan cara menyusun prompt yang **terkontrol**, **ilmiah**, dan **siap pakai** untuk permasalahan geofisika, dengan studi kasus metode geolistrik resistivitas sebagai contoh utama.


## **2. Kerangka Prompt yang Direkomendasikan: R-T-C-I-K-O**

Gunakan kerangka **R-T-C-I-K-O**. Kerangka ini sangat cocok untuk permasalahan geofisika karena memaksa AI berpikir sistematis.

| Komponen | Kepanjangan | Fungsi |
|----------|-------------|--------|
| **R** | Role | Menetapkan identitas dan kompetensi AI |
| **T** | Task | Menjelaskan tugas yang harus dikerjakan |
| **C** | Context | Memberikan latar belakang ilmiah dan konteks penelitian |
| **I** | Input | Menjelaskan data apa saja yang akan diberikan |
| **K** | Constraints | Memberikan batasan keras (satuan, larangan mengarang data, format output) |
| **O** | Output | Menentukan bentuk dan gaya jawaban yang diinginkan |

Kerangka ini mencegah AI “melamun” dan memaksa hasil yang dapat dipertanggungjawabkan secara akademik.

## **3. Contoh Prompt Lengkap (Template Siap Pakai)**

Berikut adalah prompt yang sudah dioptimasi untuk studi kasus geolistrik. Anda dapat menyalin seluruh blok di bawah ini ke AI.

```text
### PROMPT PENYELESAIAN STUDI KASUS GEOFISIKA MENGGUNAKAN METODE GEOLISTRIK

Role (R):
Anda berperan sebagai ahli geofisika yang memahami prinsip kerja, akuisisi data, pengolahan, dan interpretasi metode geolistrik resistivitas. Anda mampu melakukan perhitungan berdasarkan data hasil pengukuran geolistrik serta menjelaskan hasilnya secara akademik, sistematis, dan mudah dipahami oleh mahasiswa geofisika.

Task (T):
Selesaikan studi kasus geofisika yang menggunakan metode geolistrik berdasarkan data dan informasi yang saya berikan. Lakukan analisis mulai dari memahami kondisi dan tujuan studi kasus, mengidentifikasi konfigurasi elektroda yang digunakan, menghitung faktor geometri dan resistivitas semu, menganalisis variasi nilai resistivitas terhadap kedalaman atau lintasan, kemudian menginterpretasikan hasilnya sebagai kondisi bawah permukaan di daerah penelitian. Jelaskan setiap tahapan secara logis sehingga terlihat bagaimana data arus dan beda potensial menghasilkan informasi tentang jenis batuan, lapisan, atau keberadaan zona target seperti akuifer.

Context (C):
Metode geolistrik resistivitas mengalirkan arus listrik ke dalam bumi melalui sepasang elektroda arus, lalu mengukur beda potensial yang timbul pada sepasang elektroda potensial. Dari data tersebut diperoleh nilai resistivitas semu yang dipengaruhi oleh jenis material, porositas, kandungan fluida, salinitas, dan tingkat pelapukan batuan. Konfigurasi yang umum digunakan adalah Wenner, Schlumberger, dan dipol-dipol, dan setiap konfigurasi memiliki faktor geometri yang berbeda. Hasil pengukuran harus dikaitkan dengan tujuan penelitian, misalnya eksplorasi air tanah, pemetaan lapisan batuan, identifikasi bidang gelincir, atau pendugaan intrusi air laut.

Input (I):
Saya akan memberikan studi kasus dan data pengukuran, seperti konfigurasi elektroda, jarak antar elektroda (AB/2 dan MN/2 atau spasi a), kuat arus (I), beda potensial (ΔV), posisi lintasan, serta tabel atau gambar penampang bila tersedia. Gunakan seluruh data yang relevan. Jika ada data yang tidak lengkap, jangan mengarang nilai. Identifikasi data yang tersedia dan jelaskan hanya asumsi yang benar-benar diperlukan.

Constraints (K):
Gunakan konsep dan persamaan geolistrik yang sesuai secara ilmiah. Tentukan faktor geometri (K) sesuai konfigurasi yang digunakan, lalu hitung resistivitas semu dengan hubungan ρa = K · ΔV / I. Perhatikan satuan arus (A atau mA), beda potensial (V atau mV), jarak (m), dan resistivitas (Ωm) agar tidak terjadi kesalahan konversi. Jangan membuat atau mengasumsikan data yang tidak diberikan. Jika ada informasi yang ambigu, nyatakan asumsi yang digunakan secara eksplisit. Jika interpretasi litologi dilakukan, jelaskan bahwa satu rentang resistivitas dapat mewakili lebih dari satu jenis material, sehingga interpretasi perlu didukung informasi geologi daerah penelitian. Hubungkan hasil perhitungan dengan tujuan studi kasus. Jangan memberikan jawaban dalam bentuk tabel, daftar poin, atau format ringkasan. Seluruh hasil akhir harus berupa narasi akademik yang runtut dalam paragraf.

Output (O):
Berikan hasil penyelesaian dalam bentuk narasi akademik yang utuh dan berurutan. Narasi harus menjelaskan kondisi dan tujuan studi kasus, data yang digunakan, proses perhitungan faktor geometri dan resistivitas semu, hasil perhitungan yang diperoleh, serta interpretasi kondisi bawah permukaan dalam konteks geofisika. Sertakan nilai hasil perhitungan beserta satuannya langsung di dalam paragraf. Jelaskan makna hasil tersebut terhadap struktur lapisan, jenis material, atau potensi target yang dikaji. Gunakan bahasa Indonesia yang formal, ilmiah, dan mudah dipahami. Jangan menggunakan tabel, bullet point, numbering, atau format daftar. Seluruh output akhir hanya berupa narasi/paragraf.



---

## **4. Penjelasan Setiap Bagian Prompt (Mengapa Dirancang Seperti Itu)**

### 4.1 Role (R) — Menetapkan “Kepribadian Ilmiah”
Dengan menetapkan AI sebagai **ahli geofisika**, kita mengurangi risiko jawaban yang terlalu umum atau bersifat populer. AI akan cenderung menggunakan istilah teknis yang tepat (faktor geometri, resistivitas semu, konfigurasi elektroda, dll.).

### 4.2 Task (T) — Memberikan Alur Kerja yang Jelas
Task tidak hanya bilang “hitung resistivitas”, tetapi memaksa urutan kerja yang benar:
1. Pahami tujuan studi kasus
2. Identifikasi konfigurasi
3. Hitung K dan ρa
4. Analisis variasi
5. Interpretasi

Ini mencegah AI lompat langsung ke interpretasi tanpa perhitungan.

### 4.3 Context (C) — Memberikan Landasan Teori Singkat
Context mengingatkan AI tentang prinsip dasar metode geolistrik dan faktor-faktor yang memengaruhi resistivitas. Ini berguna agar AI tidak lupa mengaitkan hasil perhitungan dengan kondisi geologi yang mungkin terjadi.

### 4.4 Input (I) — Mengatur Ekspektasi Data
Bagian ini sangat penting. Dengan menyatakan “jangan mengarang nilai”, kita mengurangi risiko AI mengisi data yang tidak ada. AI dipaksa untuk mengakui data mana yang tersedia dan mana yang tidak.

### 4.5 Constraints (K) — Batasan Keras (Paling Krusial)
Constraints adalah bagian yang paling menentukan kualitas jawaban:
- Memaksa penggunaan rumus ρa = K · ΔV / I
- Mengingatkan konversi satuan
- Melarang tabel dan bullet point (sesuai permintaan akademik)
- Memaksa pernyataan asumsi secara eksplisit
- Mengingatkan bahwa rentang resistivitas bersifat multi-interpretasi

### 4.6 Output (O) — Mengunci Format Jawaban
Dengan meminta **narasi akademik yang utuh**, kita menghindari jawaban yang terpotong-potong. Ini sangat cocok untuk tugas laporan atau presentasi.

---

## **5. Cara Menggunakan Prompt di Praktik**

### Langkah 1: Salin Prompt R-T-C-I-K-O
Salin seluruh blok prompt di atas.

### Langkah 2: Tambahkan Data Studi Kasus
Setelah prompt, langsung tempel data pengukuran Anda. Contoh:

```text
Data pengukuran:
Konfigurasi: Wenner
Spasi a = 10 m
Arus I = 200 mA
Beda potensial ΔV = 15,4 mV
Tujuan: eksplorasi akuifer dangkal
```

### Langkah 3: Kirim ke AI
Biarkan AI mengerjakan. Jika hasilnya masih kurang memuaskan, Anda bisa menambah constraint tambahan, misalnya:

```text
Tambahan constraint:
- Hitung faktor geometri Wenner secara eksplisit
- Konversi satuan arus dan beda potensial ke satuan SI sebelum menghitung
- Sebutkan kemungkinan ambiguitas interpretasi litologi
```

### Langkah 4: Evaluasi Hasil
Periksa apakah AI:
- Menggunakan rumus yang benar
- Tidak mengarang data
- Menulis dalam bentuk narasi
- Menyebutkan satuan dengan benar
- Mengaitkan hasil dengan tujuan penelitian

## **6. Kesalahan Umum Saat Menyusun Prompt Geofisika**

| Kesalahan | Akibat | Cara Menghindari |
|-----------|--------|------------------|
| Tidak menetapkan Role | Jawaban terlalu umum | Selalu mulai dengan Role ahli geofisika |
| Task terlalu singkat | AI lompat ke interpretasi | Buat Task berisi alur kerja lengkap |
| Tidak ada Constraints satuan | Salah konversi mA ↔ A, mV ↔ V | Selalu ingatkan satuan |
| Tidak melarang mengarang data | AI mengisi nilai fiktif | Tuliskan secara eksplisit di Input & Constraints |
| Meminta tabel padahal butuh narasi | Format tidak sesuai tugas | Kunci format di bagian Output |
| Tidak mengingatkan ambiguitas litologi | Interpretasi terlalu yakin | Selalu cantumkan catatan multi-interpretasi |

## **7. Penutup**

Menyusun prompt yang baik bukan sekadar “memberi perintah”, melainkan **merancang proses berpikir** yang harus diikuti oleh AI. Dengan kerangka R-T-C-I-K-O, Anda dapat mengontrol kualitas perhitungan, kedalaman analisis, dan gaya penulisan jawaban.

Prompt yang baik akan menghasilkan:
- Perhitungan yang dapat dilacak
- Interpretasi yang hati-hati
- Narasi yang siap dimasukkan ke laporan

Gunakan modul ini sebagai referensi setiap kali Anda ingin memanfaatkan AI untuk menyelesaikan permasalahan geofisika.
